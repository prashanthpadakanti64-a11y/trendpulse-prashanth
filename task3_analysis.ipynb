```python
import pandas as pd
import numpy as np
from google.colab import files
files.upload()

# -------- Step 1: Load and Explore --------

file_path = "trends_clean.csv"

df = pd.read_csv(file_path)

print(f"Loaded data: {df.shape}")

print("\nFirst 5 rows:")
print(df.head())


# average values
avg_score = df["score"].mean()
avg_comments = df["num_comments"].mean()

print(f"\nAverage score   : {round(avg_score, 2)}")
print(f"Average comments: {round(avg_comments, 2)}")


# -------- Step 2: NumPy Analysis --------

scores = df["score"].values

print("\n--- NumPy Stats ---")

print(f"Mean score   : {round(np.mean(scores), 2)}")
print(f"Median score : {np.median(scores)}")
print(f"Std deviation: {round(np.std(scores), 2)}")

print(f"Max score    : {np.max(scores)}")
print(f"Min score    : {np.min(scores)}")


# category with most stories
top_category = df["category"].value_counts().idxmax()
top_count = df["category"].value_counts().max()

print(f"\nMost stories in: {top_category} ({top_count} stories)")


# story with most comments
max_comments_row = df.loc[df["num_comments"].idxmax()]

print(f'\nMost commented story: "{max_comments_row["title"]}" — {max_comments_row["num_comments"]} comments')


# -------- Step 3: Add New Columns --------

# engagement = comments per score
df["engagement"] = df["num_comments"] / (df["score"] + 1)

# is_popular = score > average
df["is_popular"] = df["score"] > avg_score


# -------- Step 4: Save Result --------

import os

# create data folder if not exists
os.makedirs("data", exist_ok=True)

output_file = "data/trends_analysed.csv"

df.to_csv(output_file, index=False)

print(f"\nSaved to {output_file}")
```
