```python
import pandas as pd
import matplotlib.pyplot as plt
import os
from google.colab import files
files.upload()

# loading analysed data from previous task
file_path = "trends_analysed.csv"
data_df = pd.read_csv(file_path)

print("Data loaded:", data_df.shape)

# creating output folder (if not present already)
if not os.path.exists("outputs"):
    os.makedirs("outputs")


# -------- Chart 1: Top 10 stories --------

# sorting based on score
top_data = data_df.sort_values(by="score", ascending=False).head(10)

# trimming long titles so graph looks clean
titles = []
for t in top_data["title"]:
    if len(t) > 50:
        titles.append(t[:50] + "...")
    else:
        titles.append(t)

plt.figure()

plt.barh(titles, top_data["score"])

plt.xlabel("Score")
plt.ylabel("Stories")
plt.title("Top 10 Stories by Score")

# invert so highest comes on top
plt.gca().invert_yaxis()

plt.savefig("outputs/chart1_top_stories.png")
plt.close()


# -------- Chart 2: Category counts --------

cat_counts = data_df["category"].value_counts()

plt.figure()

# using default colors (simple)
plt.bar(cat_counts.index, cat_counts.values)

plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Stories per Category")

plt.savefig("outputs/chart2_categories.png")
plt.close()


# -------- Chart 3: Scatter plot --------

plt.figure()

# separating popular and not popular
pop_data = data_df[data_df["is_popular"] == True]
non_pop = data_df[data_df["is_popular"] == False]

plt.scatter(pop_data["score"], pop_data["num_comments"], label="Popular")
plt.scatter(non_pop["score"], non_pop["num_comments"], label="Not Popular")

plt.xlabel("Score")
plt.ylabel("Comments")
plt.title("Score vs Comments")

plt.legend()

plt.savefig("outputs/chart3_scatter.png")
plt.close()


# -------- Dashboard (all charts together) --------

fig, ax = plt.subplots(1, 3, figsize=(16, 5))

# first chart again
ax[0].barh(titles, top_data["score"])
ax[0].set_title("Top Stories")
ax[0].invert_yaxis()

# second chart
ax[1].bar(cat_counts.index, cat_counts.values)
ax[1].set_title("Categories")

# third chart
ax[2].scatter(pop_data["score"], pop_data["num_comments"], label="Popular")
ax[2].scatter(non_pop["score"], non_pop["num_comments"], label="Not Popular")
ax[2].set_title("Score vs Comments")

fig.suptitle("TrendPulse Dashboard")

plt.savefig("outputs/dashboard.png")
plt.close()


print("\nAll charts created and saved in outputs folder")
```
